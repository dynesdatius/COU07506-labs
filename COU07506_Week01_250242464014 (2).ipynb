```

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

print("DSP Laboratory Environment Ready!")
```

```
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

Question 1: What is a unit impulse signal?

A unit impulse signal is a discrete-time signal that has a value of 1 at n = 0 and 0 at all other values of n.

```
n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x, basefmt="k-")
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```
n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))

axes[0].stem(n, x1, basefmt="k-")
axes[0].set_title("cos(2πn/8)")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2, basefmt="k-")
axes[1].set_title("cos(2πn/7.5)")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

Question 2: Which of the following signals is periodic: cos(2πn/8) or cos(2πn/7.5)? Explain your answer.

The signal cos(2πn/8) is periodic because its period is 8 samples, which is an integer value.

Question 4: How many samples are required to store one minute of mono audio at a sampling frequency of 48,000 Hz?

A one-minute mono audio recording requires 2,880,000 samples.
This is calculated by multiplying the sampling frequency (48,000 Hz) by 60 seconds.
Calculation:
48,000 × 60 = 2,880,000 samples.

Question 5: Why do we use a stem plot instead of a normal line plot for discrete-time signals?

Stem plots are used for discrete-time signals because they show individual samples clearly at specific time points.
They also show that discrete-time signals are not continuous.

Question 6: What happens when the sampling frequency is increased?

Increasing the sampling frequency means that more samples are taken per second, which can improve the representation of the original signal
